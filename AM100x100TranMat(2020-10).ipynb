# Loading data

```python
df = pd.read_csv(file_path+'2D_Lottery(2020-10).csv')
df.head()
```

# Data Preperation

```python
df['2D'].replace({'#N/A':np.nan, "v":np.nan, "NaN":np.nan},inplace=True)
df.dropna(inplace=True)
df.reset_index(drop=True,inplace=True)
df['2D'] = df['2D'].astype(int)
```

```python
Lot2 = df.copy()
Lot2
```

```python
Lot2 = Lot2.drop(Lot2[Lot2['AM = 1 / PM = 0'] == 0].index)
Lot2.reset_index(drop=True,inplace=True)
Lot2
```

```python
Lot2.drop(["Date", "AM = 1 / PM = 0"], axis=1, inplace=True)
Lot2.head()
```

# Train Test Split

```python
Lot2_train = Lot2[:train_end]
Lot2_test = Lot2[test_st:]
```

```python
len(Lot2_test)
```

# Training

```python
Lot2_c = Mat_2D(Lot2_train, width, n_top_prob, n_top_weight)
Lot2_c.init_prob[:5]
```

# Predict for AM

```python
def pred_next(tdy):
    pred_next = Lot2_c.pred_row([int(tdy)])
    prd = pred_next.keys()
    for i in prd:
        if i<10:
            print("0"+ str(i))
        else:
            print(i)
```
